```python
#Outliers Treatment
```

```python
df.columns
```

```python
df['origin'].value_counts()
```

```python
df['origin'] = df['origin'].map({"usa":1, "japan":2, "europe":3})
```

```python
df.info()
```

```python
df['origin'] = df['origin'].astype(int)
```

```python
df.info()
```

```python
X = df.drop("mpg", axis=1)
Y = df['mpg']
```

```python
X.isnull().sum()
```

```python
X['origin'].isnull().sum()
```

```python
Y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
X_train.shape, X_test.shape, Y_train.shape, Y_test.shape
```

```python
from sklearn.linear_model import LinearRegression
```

```python
linear_regression_model = LinearRegression()
```

```python
linear_regression_model.fit(X_train, Y_train)
```

```python
linear_regression_model.coef_
```

```python
for i, col_name in enumerate(X_train.columns):
    print(f"Coefficient of {col_name} is {linear_regression_model.coef_[i]}")
```

here the all the coeffients relatively small if one independent variable 
changes there will be not much difference in prediction
this is often called smoother model

so there feature might not be contributing in model training

```python
y_pred = linear_regression_model.predict(X_test)
```

```python
from sklearn.metrics import r2_score
```

```python
regression_r2score = r2_score(Y_test, y_pred)
print(f"The R square pf linear Regression model is {regression_r2score}")
```

REGULARIZATIOn


```python
#Regularization
```

```python
from sklearn.linear_model import Ridge

```

here alpha is nothing but the lambda value

```python
ridge_regression_model = Ridge(alpha=0.1)
```

```python
ridge_regression_model.fit(X_train, Y_train)
```

```python
for i, col_name in enumerate(X_train.columns):
    print(f"Coefficient of {col_name} is {ridge_regression_model.coef_[i]}")
```

```python
y_pred_ridge = ridge_regression_model.predict(X_test)
y_pred_ridge
```

```python
ridge_regression_r2score = r2_score(Y_test, y_pred_ridge)
print(f"The R square of linear Regression model is {ridge_regression_r2score}")
```

we dont see much variation in r2 in both mdoel

```python
from sklearn.linear_model import Lasso
```

```python
lasso_regression_model = Lasso(alpha=0.5)
lasso_regression_model
```

```python
lasso_regression_model.fit(X_train, Y_train)
```

```python
for i, col_name in enumerate(X_train.columns):
    print(f"Coefficient of {col_name} is {lasso_regression_model.coef_[i]}")
```

here 3 coffients became 0 here the lasso helps u in feature selection


```python
y_pred_lasso = lasso_regression_model.predict(X_test)
```

```python
lasso_regression_r2score = r2_score(Y_test, y_pred_lasso)
```

```python
print(f"The r2_square of the lasso model is {lasso_regression_r2score}")
```

```python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV

from sklearn.linear_model import Lasso
```

```python
param_grid1 = {"alpha":[0.01, 0.1, 0, 1, 10, 100, 1000]}
param_grid1
```

```python
lasso = Lasso()
```

```python
gridsearch_cv = GridSearchCV(estimator=lasso, param_grid=param_grid1, cv=5,
             scoring="r2", verbose=2)
```

```python
gridsearch_cv
```

```python
gridsearch_cv.fit(X_train, Y_train)
```

```python
gridsearch_cv.best_params_
```

```python
gridsearch_cv.best_estimator_
```

```python
y_pred = gridsearch_cv.best_estimator_.predict(X_test)
```

```python
r2_score(Y_test, y_pred)
```

```python
param_distribution = {"alpha":[0.0001, 0.001, 0.01, 0.1, 1, 10, 100]}
```

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
random_cv = RandomizedSearchCV(estimator=lasso, param_distributions=param_distribution, cv=5, scoring="r2", verbose=2)
```

```python
random_cv
```

```python
random_cv.fit(X_train, Y_train)
```
