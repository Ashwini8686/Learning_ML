```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_diabetes
```

```python
diabetes = load_diabetes()
```

```python
diabetes
```

```python
diabetes.feature_names
```

```python
diabetes.data
```

```python
df = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)
```

```python
df['target'] = diabetes.target
```

To predict the disease progression one year after the baseline

```python
df
```

```python
x = df.drop('target', axis = 1)
y = df['target']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
#Scaling
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
scaler
```

```python
x_train = scaler.fit_transform(X_train)
```

```python
x_test = scaler.transform(X_test)
```

```python
x_train
```

```python
x_test
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model
```

```python
model.fit(x_train, Y_train)
```

```python
df
```

```python
model.coef_
```

```python
model.intercept_
```

With 1 unit increase in the age the one year progession will increase by  1.75375799 on an average by keeping all the other features as constant

```python
y_pred = model.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, root_mean_squared_error, r2_score
```

```python
mean_absolute_error(Y_test, y_pred)
```

```python
mean_absolute_error(Y_test, y_pred)
```
