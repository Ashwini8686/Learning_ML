```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import math as m
import statsmodels.formula.api as smf

import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.datasets import make_classification
```

```python
X, Y = make_classification(n_samples=100, n_features=10, n_informative=5, n_redundant=5, n_classes=2, random_state=40)
```

```python
X
```

```python
Y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.20, random_state=42)
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
params = {"penalty":("l1", "l2", "elasticnet"),"C": [1, 2, 10, 20, 30, 40]}
```

```python
params
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model
```

```python
grid = GridSearchCV(LogisticRegression(), param_grid=params, cv=5, verbose=2)
```

```python
grid
```

```python
grid.fit(X_train, Y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
grid.best_estimator_.predict(X_test)
```

OR

```python
mod = LogisticRegression(C=1, penalty='l2')
```

```python
mod
```

```python
mod.fit(X_train, Y_train)
```

```python
y_pred = mod.predict(X_test)
```

```python
mod.predict_proba(X_test)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
print(confusion_matrix(Y_test, y_pred))
print(accuracy_score(Y_test, y_pred))
print(classification_report(Y_test, y_pred))
```
