```python
X, Y = make_classification(n_samples=1000,n_features=10, n_redundant=5, n_informative=5, n_classes=2, random_state=42)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.30, random_state=42)
```

```python
from sklearn.linear_model import  LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model
```

```python
model.fit(X_train, Y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
confusion = confusion_matrix(Y_test, y_pred)
```

```python
confusion
```

```python
accuracy = accuracy_score(Y_test, y_pred)
```

```python
from sklearn.metrics import roc_curve, auc
```

```python
y_pred_proba = model.predict_proba(X_test)[:, 1]
```

```python
y_pred_proba
```

```python
fpr, tpr, thresholds = roc_curve(Y_test, y_pred_proba)
```

```python
roc_auc = auc(fpr, tpr)
```

```python
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color="darkorange", linewidth=2, label='"ROC curve(area = %0.2f)' % roc_auc)
plt.plot([0, 1], [0, 1], color="navy", linewidth=2)
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel("FPR")
plt.ylabel("TPR")
plt.title("ROC Curve")
plt.legend(loc='lower right')
plt.show()
```

```python
#by default the prediction class is decided by 0.5, if u want ur own cutoff u have to do THE Sensitivity-specificity-tradeoff 
```

```python
from sklearn.metrics import precision_score, recall_score, accuracy_score
```

```python
thresholds = np.linspace(0, 1, 100)
precisions = []
recalls = []
accuracies = []

for i in thresholds:
    y_pred_threshold = (y_pred_proba >= i).astype(int)
    precision = precision_score(Y_test, y_pred_threshold)
    recall = recall_score(Y_test, y_pred_threshold)
    accuraci = accuracy_score(Y_test, y_pred_threshold)
    precisions.append(precision)
    recalls.append(recall)
    accuracies.append(accuraci)

plt.figure(figsize=(8, 6))
plt.plot(thresholds, precisions, label="precision")
plt.plot(thresholds, recalls, label="Recall")
plt.plot(thresholds, accuracies, label="Accuracy")

plt.xlabel("threshold Probability")
plt.ylabel("Scores")
plt.title("Precicsion, Recall, Accuracy Vs Threshold Probability")
plt.legend()
plt.grid(True)
plt.show()
```
