```python
x = df.iloc[:, :-1]
```

```python
y = df.iloc[:, -1]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(x, y, test_size=0.20, random_state=42)
```

```python
X_train.shape, X_test.shape, Y_train.shape, Y_test.shape 
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model
```

```python
model.fit(X_train, Y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
model.predict_proba(X_test) #corrsponding to higher probability class is predicted by default the cut off is 0.5
```

```python
#Evaluation matrics
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
Confusion = confusion_matrix(Y_test, y_pred)
```

```python
Confusion
```

```python
accuracy_score(Y_test, y_pred)
```

```python
print(classification_report(Y_test, y_pred))
```

| Term             | What does it tell us?                  | Why do we need it?                   |
| ---------------- | -------------------------------------- | ------------------------------------ |
| **Precision**    | How correct are predicted positives?   | Control **FP**                       |
| **Recall**       | How many actual positives did we find? | Control **FN**                       |
| **F1-score**     | Balance of Precision & Recall          | When both FP & FN matter             |
| **Support**      | Number of actual samples in each class | Understand class size/distribution   |
| **Accuracy**     | Overall correct predictions            | Quick overall performance            |
| **Macro avg**    | Equal average across classes           | Give every class equal importance    |
| **Weighted avg** | Average considering class sizes        | Reflect dataset's class distribution |


```python
from sklearn.metrics import roc_curve, auc
```

```python
#get pridicted  probabilities for one  class


y_pred_proba = model.predict_proba(X_test)[:, 1]
```

```python
y_pred_proba
```

```python
#Reciver Operating Characteristics  curve
#inputs y_test, Y_pred_proba
#it will return TPR, FPR with different cutoff of probability


fpr, tpr, thresholds = roc_curve(Y_test, y_pred_proba)
```

```python
fpr
```

```python
tpr
```

```python
thresholds
```

```python
roc_auc = auc(fpr, tpr)
```

```python
roc_auc
```

```python
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color="darkorange", linewidth=2, label='"ROC curve(area = %0.2f)' % roc_auc)
plt.plot([0, 1], [0, 1], color="navy", linewidth=2)
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel("FPR")
plt.ylabel("TPR")
plt.title("ROC Curve")
plt.legend(loc='lower right')
plt.show()
```
