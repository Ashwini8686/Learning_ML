```python
#outlier detection and treatment
```

```python
df.columns
```

```python
st_grade = df.copy()
```

```python
st_grade
```

```python
def detect_outliers(columns):
    outliers = []

    for column in columns:
        Q1 = st_grade[column].quantile(0.25)
        Q3 = st_grade[column].quantile(0.75)
        IQR = Q3 - Q1

        mask = (
            (st_grade[column] >= Q1 - 1.5 * IQR) &
            (st_grade[column] <= Q3 + 1.5 * IQR)
        )

        mask = mask.to_numpy()

        false_index = np.argwhere(~mask)

        outliers.append(false_index)

    return np.unique(np.concatenate(outliers).ravel())
```

```python
num_col = ["age", "absences"]
detect_outliers(num_col)
```
