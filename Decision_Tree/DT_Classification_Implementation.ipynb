```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import math as m
import statsmodels.formula.api as smf

import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.datasets import load_iris
```

```python
data = load_iris()
```

```python
print(data.DESCR)
```

```python
data.data
```

```python
data.feature_names
```

```python
df = pd.DataFrame(data.data, columns=data.feature_names)
```

```python
df
```

```python
df['target'] = data.target
```

```python
df
```

```python
X = df.iloc[:, :-1]
Y = df.iloc[:, -1]
```

```python
X
```

```python
Y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.30, random_state=42)
```

```python
X_train.shape, X_test.shape, Y_train.shape, Y_test .shape
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model = DecisionTreeClassifier(criterion='entropy')
model
```

```python
model.fit(X_train, Y_train)
```

```python
from sklearn import tree
plt.figure(figsize=(12, 10))
tree.plot_tree(model, filled=True)
plt.show()
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(Y_test, y_pred)
```

Post Pruning

After making DT u have to cut it to avoid Overfitting

by observing the DT Plot we got to know that it is actually overfitting for the last split it still deviding the data in such a way that each node is 100% pure
this is only leading to very low data points

```python
model = DecisionTreeClassifier(criterion='entropy', max_depth=3)
model
```

```python
model.fit(X_train, Y_train)
```

```python
from sklearn import tree
plt.figure(figsize=(12, 10))
tree.plot_tree(model, filled=True)
plt.show()
```

```python
y = model.predict(X_test)
```

```python
accuracy_score(Y_test, y)
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(Y_test, y))
```

Pre-Pruning by Using Hyperparameter Tuining

```python
parameter = {
    "criterion":['gini', 'entropy', 'log_loss'], 
    "splitter": ['best', 'random'], 
    "max_depth": [1, 2, 3, 4, 5, 6],
    "max_features":['sqrt', 'log2', 'auto']
}
```

```python
from sklearn.model_selection import GridSearchCV
dt = DecisionTreeClassifier()
```

```python
mod = GridSearchCV(dt, param_grid=parameter, cv=5, scoring='accuracy', verbose=2)
```

```python
mod.fit(X_train, Y_train)
```

```python
mod.best_params_
```

```python
mod.best_score_
```

```python
y_p = mod.predict(X_test)
```

```python
accuracy_score(Y_test, y_p)
```
